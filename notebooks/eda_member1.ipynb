```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1 — Correct path to file (because your notebook is in 'notebooks/')
file_path = "../data/features/url_features.csv"

# Step 2 — Check and load
if not os.path.exists(file_path):
    raise FileNotFoundError("url_features.csv not found — please make sure it's in ../data/features/")
else:
    print("✅ Found file at:", file_path)

# Step 3 — Load dataset
df = pd.read_csv(file_path, encoding="latin1", on_bad_lines="skip")
print("✅ File loaded successfully!")
print("Shape:", df.shape)

# Step 4 — Preview the data
display(df.head())

# Step 5 — Summary statistics
print("\n📊 Summary statistics:")
display(df.describe())

# Step 6 — Correlation heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap — URL Features")
plt.show()

# Step 7 — URL Length Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["url_length"], bins=50, kde=True)
plt.title("Distribution of URL Lengths")
plt.xlabel("URL Length")
plt.ylabel("Frequency")
plt.show()

# Step 8 — HTTPS vs Non-HTTPS
plt.figure(figsize=(6,4))
sns.countplot(x="has_https", data=df)
plt.title("HTTPS vs Non-HTTPS URLs")
plt.xlabel("Has HTTPS (1=Yes, 0=No)")
plt.ylabel("Count")
plt.show()
```
