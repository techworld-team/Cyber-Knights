```python
import pandas as pd

# ✅ Adjust path if needed (use "../" if your notebook is inside notebooks/)
file_path = "../data/features/email_features.csv"

# Load only a subset of columns for faster analysis
usecols = ['num_words', 'num_chars', 'has_link']  # update if your file has other names
df = pd.read_csv(file_path, encoding="latin1", on_bad_lines="skip", usecols=usecols)

print("✅ File loaded successfully!")
print(df.info())
df.head()
```

```python
df.describe(include='all')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df['num_words'], bins=50, kde=True)
plt.title("Email Word Count Distribution")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Feature Correlation Matrix")
plt.show()
```

```python
# -----------------------------------------------
# Member 1 - URL Feature Exploration
# -----------------------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Step 1 — Verify file path
url_path = "data/features/url_features.csv"
if not os.path.exists(url_path):
    raise FileNotFoundError(f"❌ File not found at {url_path}")

# Step 2 — Load dataset
print("📥 Loading URL features...")
df = pd.read_csv(url_path, encoding="latin1", on_bad_lines="skip")
print("✅ Data loaded successfully!")

# Step 3 — Quick overview
display(df.head())
print("\nDataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())

# Step 4 — Descriptive stats
display(df.describe())

# Step 5 — Missing values check
print("\nMissing values per column:")
print(df.isnull().sum())

# Step 6 — Plot some insights
sns.set(style="whitegrid")

plt.figure(figsize=(8,5))
sns.histplot(df["url_length"], bins=50, kde=True)
plt.title("Distribution of URL Lengths")
plt.xlabel("URL Length")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8,5))
sns.countplot(x="has_https", data=df)
plt.title("HTTPS Presence in URLs")
plt.xticks([0, 1], ["No HTTPS", "Has HTTPS"])
plt.show()

plt.figure(figsize=(8,5))
sns.countplot(x="has_ip", data=df)
plt.title("URLs Containing IP Address")
plt.xticks([0, 1], ["No IP", "Has IP"])
plt.show()
```
